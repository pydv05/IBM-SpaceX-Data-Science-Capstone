# Exploratory Data Analysis and Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
df["Class"] = df["Outcome"].astype(str).str.startswith("True").astype(int)
df["PayloadMass"] = pd.to_numeric(df["PayloadMass"], errors="coerce")
```

```python
# Launch count by site
site_counts = df["LaunchSite"].value_counts()
display(site_counts)

site_counts.plot(kind="bar", title="Launches by Launch Site")
plt.xlabel("Launch Site")
plt.ylabel("Number of Launches")
plt.tight_layout()
plt.show()
```

```python
# Landing success rate by launch site
site_rates = df.groupby("LaunchSite")["Class"].mean().sort_values(ascending=False)
display((site_rates * 100).round(1))

site_rates.mul(100).plot(kind="bar", title="Landing Success Rate by Launch Site")
plt.ylabel("Success rate (%)")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
# Payload versus flight number, separated by landing outcome
sns.scatterplot(data=df, x="FlightNumber", y="PayloadMass", hue="Class")
plt.title("Payload Mass vs Flight Number")
plt.tight_layout()
plt.show()
```

```python
# Success trend over time
year_rates = df.groupby("Year")["Class"].mean().mul(100)
display(year_rates)

year_rates.plot(marker="o", title="Landing Success Rate by Year")
plt.ylabel("Success rate (%)")
plt.ylim(0, 100)
plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()
```

Key interpretation to compare with the final report: landing success improves over time, and the KSC/VAFB sites show strong success rates in the course-era sample.
