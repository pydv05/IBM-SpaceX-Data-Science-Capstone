# SpaceX Data Collection — REST API

This notebook collects historical SpaceX launches from the public SpaceX REST API, normalizes nested JSON, filters Falcon 9 launches, and saves a clean raw collection. The workflow follows the IBM Applied Data Science Capstone methodology.

```python
import requests
import pandas as pd
import json

API_URL = "https://api.spacexdata.com/v4/launches/past"
response = requests.get(API_URL, timeout=30)
response.raise_for_status()
launches = response.json()

df_api = pd.json_normalize(launches)
print("Total historical launches returned:", len(df_api))
df_api.head()
```

```python
# Keep Falcon 9 missions and inspect the normalized structure.
falcon9 = df_api[df_api["rocket"].notna()].copy()
print("Normalized columns:", len(df_api.columns))
print("Rows:", len(df_api))
df_api.filter(regex="^(name|date_utc|success|rocket|launchpad)$").head()
```
