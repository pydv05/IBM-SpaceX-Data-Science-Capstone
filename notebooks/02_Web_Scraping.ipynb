# SpaceX Data Collection — Wikipedia Web Scraping

Collect supplementary launch-table information from the SpaceX launch history on Wikipedia using pandas HTML-table extraction. Inspect the tables and select the launch-history table for downstream comparison.

```python
import pandas as pd

WIKI_URL = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
tables = pd.read_html(WIKI_URL)
print("Tables discovered:", len(tables))
for i, t in enumerate(tables[:10]):
    print(i, t.shape, list(t.columns)[:5])
```

```python
# Choose tables containing launch/date information.
candidates = []
for i, t in enumerate(tables):
    text = " ".join(map(str, t.columns)).lower()
    if "date" in text or "launch" in text:
        candidates.append((i, t.shape))
candidates[:10]
```
