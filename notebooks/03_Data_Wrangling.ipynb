# SpaceX Data Wrangling and Landing-Class Creation

```python
import pandas as pd
import numpy as np

URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"
df = pd.read_csv(URL)
print(df.shape)
display(df.head())
```

```python
# Standardize dates and create the binary landing target.
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Year"] = df["Date"].dt.year

# IBM capstone convention: any outcome beginning with True is a successful landing.
df["Class"] = df["Outcome"].astype(str).str.startswith("True").astype(int)

# Normalize common missing values.
df["PayloadMass"] = pd.to_numeric(df["PayloadMass"], errors="coerce")
df["PayloadMass"] = df["PayloadMass"].fillna(df["PayloadMass"].median())

print(df["Class"].value_counts())
display(df[["FlightNumber","Date","LaunchSite","Outcome","PayloadMass","Class"]].head())
```

```python
# Save a local cleaned copy for optional offline use.
df.to_csv("../data/cleaned_spacex_launches.csv", index=False)
print("Saved ../data/cleaned_spacex_launches.csv")
```

Course-analysis note: the historical capstone dataset contains 90 records and a binary Class target. The final report used this cleaned representation for EDA and classification.
