# Interactive Launch-Site Map with Folium

```python
import pandas as pd
import folium

df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
df["Class"] = df["Outcome"].astype(str).str.startswith("True").astype(int)

# One coordinate per operational launch site.
sites = (df.groupby(["LaunchSite", "Longitude", "Latitude"], as_index=False)
           .agg(launches=("Class","size"), success_rate=("Class","mean")))
sites["success_rate"] *= 100
sites
```

```python
m = folium.Map(location=[32.5, -100], zoom_start=4)

for _, row in sites.iterrows():
    popup = (
        f"<b>{row['LaunchSite']}</b><br>"
        f"Launches: {int(row['launches'])}<br>"
        f"Success rate: {row['success_rate']:.1f}%"
    )
    folium.Marker(
        [row["Latitude"], row["Longitude"]],
        popup=popup,
        tooltip=row["LaunchSite"]
    ).add_to(m)

m.save("../data/spacex_launch_sites.html")
m
```

The interactive map connects launch outcomes with geography. The major operational sites are located on the Florida and California coasts, consistent with the final project report.
