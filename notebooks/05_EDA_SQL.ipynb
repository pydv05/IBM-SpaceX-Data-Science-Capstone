# Exploratory Data Analysis with SQL

```python
import pandas as pd
import sqlite3

df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv")
conn = sqlite3.connect("../data/spacex.db")
df.to_sql("SPACEXTBL", conn, if_exists="replace", index=False)
print("Rows loaded:", pd.read_sql_query("SELECT COUNT(*) AS n FROM SPACEXTBL", conn).iloc[0,0])
```

### Core capstone SQL queries

```python
queries = {
    "First 20 records":
        'SELECT * FROM SPACEXTBL LIMIT 20',
    "Minimum payload":
        'SELECT MIN(PayloadMass) AS Minimum_Payload_Mass FROM SPACEXTBL',
    "Total payload":
        'SELECT SUM(PayloadMass) AS Total_Payload_Mass FROM SPACEXTBL',
    "Mission outcomes by launch site":
        'SELECT LaunchSite, COUNT(Outcome) AS MISSION_OUTCOME_COUNT FROM SPACEXTBL GROUP BY LaunchSite',
    "Unique launch sites":
        'SELECT DISTINCT LaunchSite FROM SPACEXTBL'
}

for title, sql in queries.items():
    print("\n---", title, "---")
    display(pd.read_sql_query(sql, conn))
```

```python
# Additional success-rate query
sql = '''
SELECT LaunchSite,
       COUNT(*) AS launches,
       SUM(Class) AS successful_landings,
       ROUND(100.0 * AVG(Class), 1) AS success_rate_percent
FROM SPACEXTBL
GROUP BY LaunchSite
ORDER BY success_rate_percent DESC
'''
display(pd.read_sql_query(sql, conn))
```
